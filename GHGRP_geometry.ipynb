```python
import pandas
import geopandas

from google import colab
from google.colab import drive
drive.mount('/content/drive')
```

```python
ghg_states = pandas.read_csv("/content/drive/Shareddrives/EDGI - Shared NEW/12_Environmental_Enforcement_Watch/09_Data/Climate_Data_Project/web_data/aggregated_emissions_by_state_2023.csv")
```

```python
import requests
import zipfile
import io

url = "https://www2.census.gov/geo/tiger/TIGER2025/STATE/tl_2025_us_state.zip"

# Download
r = requests.get(url)

# Extract
with zipfile.ZipFile(io.BytesIO(r.content)) as z:
    z.extractall("states_shapefile")
```

```python
states = geopandas.read_file("states_shapefile/tl_2025_us_state.shp")
```

```python
ghg_states
```

```python
states
```

```python
ghgstates_geo = pandas.merge(ghg_states, states, left_on="Effective State", right_on="STUSPS")
```

```python
ghgstates_geo = geopandas.GeoDataFrame(ghgstates_geo, geometry = "geometry")
```

```python
ghgstates_geo.to_file("ghg_state.shp")
```

County

```python
ghg_county = pandas.read_csv("https://github.com/edgi-govdata-archiving/GHG-CDP/raw/refs/heads/main/output_files/aggregated_emissions_by_county.csv")
```

```python
import requests
import zipfile
import io

url = "https://www2.census.gov/geo/tiger/TIGER2025/COUNTY/tl_2025_us_county.zip" #Have to use 2020 here because https://www.zip-codes.com/connecticut-change-to-planning-regions.asp
#GHGRP still reported CT County names with 2023 data. Theoretically, we could/should use facility address to map CT emissions to the correct planning region.

# Download
r = requests.get(url)

# Extract
with zipfile.ZipFile(io.BytesIO(r.content)) as z:
    z.extractall("county_shapefile")
```

```python
counties = geopandas.read_file("county_shapefile/tl_2025_us_county.shp")
```

```python
ghg_county
```

```python
counties
```

```python
ghg_county_2023 = ghg_county[ghg_county["Year"] == 2023]
```

```python
ghg_county_2023
```

```python
ghg_county_2023["County_FIPS"] = ghg_county_2023["County_FIPS"].astype(str)
```

```python
#Fix leading 0s
ghg_county_2023['County_FIPS'] = ghg_county_2023['County_FIPS'].str.zfill(5)
```

```python
ghg_county_2023[ghg_county_2023["COUNTYNAME_cleaned"] == "Fairfield"]["County_FIPS"]
```

```python
ghg_county_2023
```

```python
ghgcounty_geo = pandas.merge(ghg_county_2023, counties, left_on="County_FIPS", right_on="GEOID")
```

```python
ghgcounty_geo
```

```python
missing_items = ghg_county_2023[~ghg_county_2023['County_FIPS'].isin(ghgcounty_geo['County_FIPS'])]
missing_items
```

```python
missing_items_2 = counties[~counties['GEOID'].isin(ghg_county_2023['County_FIPS'])]
missing_items_2
```

```python
ghgcounty_geo = geopandas.GeoDataFrame(ghgcounty_geo, geometry = "geometry")
```

```python
ghgcounty_geo.to_file("ghg_county_2023.shp")
```
