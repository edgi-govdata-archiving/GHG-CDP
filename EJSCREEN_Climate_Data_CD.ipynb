# EJ Screen Climate Change Indicators

This notebook is intended to respond to the question "Which communities are most at risk for the impacts of climate change?" using data from EJScreen v2.3 (2024). Using the climate change indicators data, air pollution data from EJScreen's core environmental indicators, and the demographic and supplemental demographic indices, we identified block groups in the top twenty percent (≥80th percentile) nationally for flood, wildfire, and extreme heat, and air pollution indicators, *and* either of the two social vulnerability indices (the demographic index or the supplemental demographic index). For greater consistency across data sources, we focused on data from the contiguous US. This was completed in "EJSCREEN Climate Data Merge.ipynb".

In this notebook, we calculate the percentage of blocks within each CD that meet these criteria, sum the populations implicated, and visualize the results in tables, maps, and charts.

Sum the total CD populations from "POP100" census data

```python
ejscreen_cd['POP100_SUM'] = ejscreen_cd.groupby('GEOID_CD')['POP100'].transform('sum')
```

Sum the vulnerable population within each CD based on the criteria above ***NOTE: POP100 comes from the CD lookup file data, so should be the right field to use to sum populations from blocks up to CDs.***  
We want to do this before CDs becomes our index or primary unit of analysis, while we still have the block data.

```python
ejscreen_cd['airquality_vulnerable_pop'] = (
    ejscreen_cd['POP100']
    .where(ejscreen_cd['airquality_demog_high'] == 1, 0)
    .groupby(ejscreen_cd['GEOID_CD'])
    .transform('sum')
)

ejscreen_cd['flood_vulnerable_pop'] = (
    ejscreen_cd['POP100']
    .where(ejscreen_cd['flood_demog_high'] == 1, 0)
    .groupby(ejscreen_cd['GEOID_CD'])
    .transform('sum')
)
ejscreen_cd['wildfire_vulnerable_pop'] = (
    ejscreen_cd['POP100']
    .where(ejscreen_cd['wildfire_demog_high'] == 1, 0)
    .groupby(ejscreen_cd['GEOID_CD'])
    .transform('sum')
)
ejscreen_cd['heat_vulnerable_pop'] = (
    ejscreen_cd['POP100']
    .where(ejscreen_cd['heat_demog_high'] == 1, 0)
    .groupby(ejscreen_cd['GEOID_CD'])
    .transform('sum')
)
```

```python
ejscreen_cd['flood_vulnerable_pop_30'] = (
    ejscreen_cd['POP100']
    .where(ejscreen_cd['flood_demog_high_30'] == 1, 0)
    .groupby(ejscreen_cd['GEOID_CD'])
    .transform('sum')
)
ejscreen_cd['wildfire_vulnerable_pop_30'] = (
    ejscreen_cd['POP100']
    .where(ejscreen_cd['wildfire_demog_high_30'] == 1, 0)
    .groupby(ejscreen_cd['GEOID_CD'])
    .transform('sum')
)
```

```python
ejscreen_cd['airquality_pop_pct_cd'] = ejscreen_cd['airquality_vulnerable_pop'] / ejscreen_cd['POP100_SUM'] * 100

ejscreen_cd['flood_pop_pct_cd'] = ejscreen_cd['flood_vulnerable_pop'] / ejscreen_cd['POP100_SUM'] * 100

ejscreen_cd['wildfire_pop_pct_cd'] = ejscreen_cd['wildfire_vulnerable_pop'] / ejscreen_cd['POP100_SUM'] * 100

ejscreen_cd['heat_pop_pct_cd'] = ejscreen_cd['heat_vulnerable_pop'] / ejscreen_cd['POP100_SUM'] * 100
```

```python
ejscreen_cd['flood_30yr_pop_pct_cd'] = ejscreen_cd['flood_vulnerable_pop_30'] / ejscreen_cd['POP100_SUM'] * 100

ejscreen_cd['wildfire_30yr_pop_pct_cd'] = ejscreen_cd['wildfire_vulnerable_pop_30'] / ejscreen_cd['POP100_SUM'] * 100
```

```python
#Select which columns to keep
cd_columns = [
    "GEOID_CD",
    "STATE_NAME",
    "POP100_SUM",
    #"airquality_high_block_cd_count",
    "airquality_vulnerable_pop",
    "flood_vulnerable_pop",
    "wildfire_vulnerable_pop",
    "heat_vulnerable_pop",
    #"total_cd_block",
    #"flood_demog_high_block_cd_count",
    #"wildfire_demog_high_block_cd_count",
    #"heat_demog_high_block_cd_count",
    #"airquality_pct_cd",
    #"flood_pct_cd",
    #"wildfire_pct_cd",
    #"heat_pct_cd",
    'flood_vulnerable_pop_30',
    'wildfire_vulnerable_pop_30',
    #"flood_demog_high_block_cd_count_30",
    #"wildfire_demog_high_block_cd_count_30",
    #"flood_pct_cd_30",
    #"wildfire_pct_cd_30",
    "airquality_pop_pct_cd",
    "flood_pop_pct_cd",
    "wildfire_pop_pct_cd",
    "heat_pop_pct_cd",
    "flood_30yr_pop_pct_cd",
    "wildfire_30yr_pop_pct_cd"
]
#and subset so that we're just looking at one row for each CD
ej_cc_cd = ejscreen_cd[cd_columns].drop_duplicates(subset="GEOID_CD")
```

#Combined Map

```python
#Reduce file size if possible
# Keep only fields needed for map + tooltip
map_cols = [
    "STATE_NAME_y",
    "CDFIPS",
    "NAME",
    "PARTY",

    # Heat
    "heat_pop_pct_cd",
    "heat_vulnerable_pop",


    # Wildfire
    "wildfire_pop_pct_cd",
    "wildfire_vulnerable_pop",


    # Flood
    "flood_pop_pct_cd",
    "flood_vulnerable_pop",


    # Air Pollution
    "airquality_pop_pct_cd",
    "airquality_vulnerable_pop",


    "geometry"
]


cd_map = ej_cc_cd[map_cols].copy()


# Make sure CRS is web-friendly
cd_map = cd_map.to_crs(epsg=4326)


# Simplify geometry
# This reduces vertices while preserving district boundaries
cd_map["geometry"] = cd_map.geometry.simplify(
    tolerance=0.02,
    preserve_topology=True
)


# Reset index (prevents extra index metadata)
cd_map = cd_map.reset_index(drop=True)
```

```python
states_simple = states.copy()

states_simple["geometry"] = states_simple.geometry.simplify(
    tolerance=0.02,
    preserve_topology=True
)
```

```python
from branca.element import Element


m = folium.Map(
    location=[40.5, -100],
    zoom_start=4,
    min_zoom=3,
    max_zoom=8,
    max_bounds=True,
    tiles=None
)

folium.TileLayer(
    "ESRI.worldgraycanvas",
    control=False
).add_to(m)


# --------------------------------
# State boundary pane
# --------------------------------

folium.map.CustomPane(
    "statePane",
    z_index=650
).add_to(m)


# --------------------------------
# State boundaries
# --------------------------------

folium.GeoJson(
    states_simple,
    style_function=lambda feature: {
        "fillColor": "transparent",
        "color": "black",
        "opacity": 0.6,
        "weight": 0.5,
        "fillOpacity": 0
    },
    control=False,
    pane="statePane",
    interactive=False
).add_to(m)


# --------------------------------
# Function to add CD layer
# --------------------------------

def add_layer(
    map_object,
    dataframe,
    layer_name,
    value_col,
    vulnerable_pop_col,
    colors,
    show=False
):

    colormap = cm.LinearColormap(
        colors=colors,
        vmin=dataframe[value_col].min(),
        vmax=dataframe[value_col].max()
    )


    def style_function(feature):

        value = feature["properties"].get(value_col)

        if value is None:

            return {
                "fillColor": "#d9d9d9",
                "color": "black",
                "weight": 0.5,
                "fillOpacity": 1
            }

        return {
            "fillColor": colormap(value),
            "color": "black",
            "weight": 0.5,
            "fillOpacity": 1
        }


    fg = folium.FeatureGroup(
        name=layer_name,
        overlay=False,
        show=show
    )


    folium.GeoJson(
        dataframe,
        style_function=style_function,
        tooltip=folium.GeoJsonTooltip(
            fields=[
                "STATE_NAME_y",
                "CDFIPS",
                value_col,
                vulnerable_pop_col,
                "PARTY"
            ],
            aliases=[
                "State:",
                "District:",
                "% of District Population Most Vulnerable:",
                "Estimated Vulnerable Population:",
                "Party (119th Congress):"
            ],
            localize=True,
            sticky=True
        )
    ).add_to(fg)


    fg.add_to(map_object)


# --------------------------------
# Define layers
# --------------------------------

layers = [

    {
        "name": "Extreme Heat",
        "value": "heat_pop_pct_cd",
        "population": "heat_vulnerable_pop",
        "colors": ["whitesmoke", "darkred"],
        "legend_label": "% Population Most at Risk from Extreme Heat and Socially Vulnerable",
        "show": True
    },

    {
        "name": "Wildfire",
        "value": "wildfire_pop_pct_cd",
        "population": "wildfire_vulnerable_pop",
        "colors": ["#F9F7F4", "#C97C08"],
        "legend_label": "% Population Most at Risk from Wildfire and Socially Vulnerable",
        "show": False
    },

    {
        "name": "Flooding",
        "value": "flood_pop_pct_cd",
        "population": "flood_vulnerable_pop",
        "colors": ["#F9F7F4", "#19649F"],
        "legend_label": "% Population Most at Risk from Flooding and Socially Vulnerable",
        "show": False
    },

    {
        "name": "Air Pollution",
        "value": "airquality_pop_pct_cd",
        "population": "airquality_vulnerable_pop",
        "colors": ["#F9F7F4", "#53ADA4"],
        "legend_label": "% Population Most at Risk from Air Pollution and Socially Vulnerable",
        "show": False
    }

]


# --------------------------------
# Build layers
# --------------------------------

for layer in layers:

    add_layer(
        map_object=m,
        dataframe=cd_map,
        layer_name=layer["name"],
        value_col=layer["value"],
        vulnerable_pop_col=layer["population"],
        colors=layer["colors"],
        show=layer["show"]
    )


# --------------------------------
# Methodology / Legend panel
# --------------------------------

legend_html = """

<div id="info_panel"
style="
position: fixed;
bottom: 30px;
left: 30px;
z-index:9999;
background:white;
padding:10px;
border:2px solid grey;
border-radius:5px;
font-size:14px;
width:360px;
max-height:420px;
overflow-y:auto;
">


<!-- ========================= -->
<!-- Collapse button -->
<!-- ========================= -->

<button
onclick="
var panel = document.getElementById('info_content');
var arrow = document.getElementById('info_arrow');

if (panel.style.display === 'none') {
    panel.style.display = 'block';
    arrow.innerHTML = '▼';
} else {
    panel.style.display = 'none';
    arrow.innerHTML = '▲';
}
"
style="
position:absolute;
top:4px;
right:4px;
width:30px;
height:30px;
border:none;
background:none;
font-size:20px;
font-weight:bold;
cursor:pointer;
padding:0;
line-height:30px;
">

<span id="info_arrow">▲</span>

</button>


<!-- ========================= -->
<!-- Methodology / Legend tabs -->
<!-- ========================= -->

<div style="
display:flex;
margin-bottom:10px;
padding-right:35px;
">

<button
onclick="
document.getElementById('info_content').style.display='block';
document.getElementById('methodology').style.display='block';
document.getElementById('legend').style.display='none';
document.getElementById('info_arrow').innerHTML='▼';
"
style="
flex:1;
padding:7px;
font-weight:bold;
cursor:pointer;
margin-right:4px;
">
Methodology
</button>


<button
onclick="
document.getElementById('info_content').style.display='block';
document.getElementById('methodology').style.display='none';
document.getElementById('legend').style.display='block';
document.getElementById('info_arrow').innerHTML='▼';
"
style="
flex:1;
padding:7px;
font-weight:bold;
cursor:pointer;
">
Legend
</button>

</div>


<!-- ========================= -->
<!-- Collapsible content -->
<!-- ========================= -->

<div id="info_content" style="display:none;">



<!-- ========================= -->
<!-- Methodology -->
<!-- ========================= -->

<div id="methodology">

<b>Methodology</b>

<p>

This map estimates the percentage of each Congressional
District's population living in Census blocks that
simultaneously meet:

<ul>

<li>High climate hazard risk (top 20% nationally)</li>

<li>High demographic vulnerability (top 20% nationally)</li>

</ul>

<br>


<div style="
border:0px;
height:160px;
display:flex;
align-items:center;
justify-content:center;
margin-top:10px;
">

<img src="https://in2the4est.github.io/TQR/Screenshot%202026-08-20%20110643.png"
     style="width:100%;">

</div>

<br>
<br>
<br>

<p>


Data for this tool were obtained from the Environmental Justice
Screening and Mapping Tool (Version 2.3) or EJScreen.

<a href="https://pedp-ejscreen.azurewebsites.net/"
target="_blank">
EJScreen
</a>

is maintained by

<a href="https://publicenvirodata.org/"
target="_blank">
Public Environmental Data Partners
</a>.

More information about these indicators is available from the

<a href="https://www.epa.gov/system/files/documents/2024-07/ejscreen-tech-doc-version-2-3.pdf"
target="_blank">
EJScreen technical documentation
</a>.

<p>

This map focuses on the contiguous US for greater consistency
across indicators with varying data availability.

</p>


<b>Climate Change Indicators</b>

<p>

<b>Flood Risk:</b> The household risk of flooding under 2022
weather conditions as modeled by First Street. Includes
percentiles for conditions projected to exist in 30 years.

</p>

<p>

<b>Wildfire Risk:</b> The household risk of wildfire exposure
under 2022 weather conditions as modeled by First Street.
Includes percentiles for conditions projected to exist in
30 years.

</p>

<p>

<b>*Extreme Heat:</b> The number of days over 90º F
(from 2019-2023), including annual totals, averages for the
entire period, and maximums for the entire period per block
group. This is calculated from PRISM Climate Group's gridded
daily Tmax estimates. Data was derived by EPA from PRISM
Climate Group.

</p>

<p>

<b>*For this study:</b> We modified the Extreme Heat indicator
by calculating national extreme heat percentile rankings based
on each block group's highest annual count of days above 90°F
between 2019 and 2023.

</p>


<b>Air Pollution Index</b>

<p>

<b>Particulate Matter 2.5:</b> The PM 2.5 indicator is a measure
of potential exposure to inhalable particles that are 2.5
micrometers or smaller. This is measured in terms of annual
average concentration in air (2020).

</p>

<p>

<b>Ozone:</b> Annual mean top 10 of daily maximum 8-hour
concentration in air. A measure of potential exposure to
ground-level ozone (2020).

</p>

<p>

<b>Nitrogen Dioxide (NO2):</b> Average annual nitrogen dioxide
(NO2) levels expressed as part per billion (by volume) (2020).

</p>

<p>

<b>Diesel Particulate Matter:</b> Diesel particulate matter
level in air (2020).

</p>

<p>

<b>Toxic releases to air:</b> Risk Screening Environmental
Indicator (RSEI) modeled toxicity-weighted concentrations in
air of Toxic Release Inventory (TRI) listed chemicals (2021).

</p>

<p>

<b>Traffic proximity and volume:</b> Count of vehicles
(annual average daily traffic [AADT]) at major roads within
10 kilometers (km), divided by distance in km (2020).

</p>


<b>Social Vulnerability Index</b>

<p>

We created a social vulnerability measure comprised of two
indices from EJScreen: the demographic index and the
supplemental demographic index. EJScreen calculated these
indices as follows:

</p>

<p>

The demographic index is a standardized score calculated from
Census data regarding the percent people of color and percent
low income for each block group as identified by the American
Communities Survey (ACS) 2018-2022 five year summaries.

</p>

<p>

The supplemental demographic index also includes percent low
income as a factor, and additionally incorporates percentages
of block group populations identified by the ACS as low life
expectancy, persons with disabilities, limited English speaking
proficiency, and less than high-school education.

Similar to their climate change indicator counterparts, these
standardized scores are then used to calculate national
percentile rankings of census block groups for each index.

</p>

<p>

This map aggregates EJScreen block group data to the
congressional district level by first assigning EJScreen data
from block groups to the smaller geographic unit referred to
as blocks, and then aggregates up to the congressional district
level from there.

</p>

<p>

The goal of this map is to provide a resource for communicating
with publicly elected officials about the impacts of climate
change on their constituents.

This map is not intended to inform specific policy, legal,
insurance, or other consequential decisions.

For more information about this tool, including analysis and
limitations, visit the corresponding report here [add link].

</p>

</div>


<!-- ========================= -->
<!-- Legend -->
<!-- ========================= -->

<div id="legend" style="display:none;">

"""


# --------------------------------
# Add legend entries
# --------------------------------

for layer in layers:

    col = layer["value"]

    min_val = cd_map[col].min()
    max_val = cd_map[col].max()

    color1 = layer["colors"][0]
    color2 = layer["colors"][1]

    legend_html += f"""

    <br>

    <b>{layer['name']}</b><br>

    {layer['legend_label']}

    <div style="
    width:200px;
    height:15px;
    background:linear-gradient(
        to right,
        {color1},
        {color2}
    );
    ">
    </div>

    <div style="
    display:flex;
    justify-content:space-between;
    width:200px;
    ">
        <span>{min_val:.1f}%</span>
        <span>{max_val:.1f}%</span>
    </div>

    """


# --------------------------------
# Close legend, content, and panel
# --------------------------------

legend_html += """

</div> <!-- legend -->

</div> <!-- info_content -->

</div> <!-- info_panel -->

"""


m.get_root().html.add_child(
    Element(legend_html)
)


# --------------------------------
# Layer control
# --------------------------------

folium.LayerControl(
    collapsed=False
).add_to(m)


# --------------------------------
# Map title
# --------------------------------

title_html = """
<div style="
position: fixed;
top: 10px;
left: 50%;
transform: translateX(-50%);
z-index:9999;
background-color: white;
padding: 8px 16px;
border: 2px light gray;
border-radius: 5px;
max-width: 600px;
width: 100%;
font-size: 18px;
font-weight: bold;
text-align: center;
box-shadow: 2px 2px 6px rgba(0,0,0,0.3);
">

Percent of People Living in Areas Most Vulnerable to Climate
Change Impacts by Congressional District

</div>
"""

m.get_root().html.add_child(
    Element(title_html)
)


# --------------------------------
# Display map
# --------------------------------

m
```
