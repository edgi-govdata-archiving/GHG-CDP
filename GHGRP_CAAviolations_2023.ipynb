```python
import pandas as pd
from google.colab import drive
import geopandas as gpd
drive.mount('/content/drive')
```

```python
ghgrp = pd.read_csv("/content/drive/Shareddrives/EDGI - Shared NEW/12_Environmental_Enforcement_Watch/09_Data/Climate_Data_Project/facility_frsid_metadata.csv")
violations = pd.read_csv("/content/drive/Shareddrives/EDGI - Shared NEW/12_Environmental_Enforcement_Watch/09_Data/Climate_Data_Project/ICIS-AIR_VIOLATION_HISTORY-2023.csv")
frs_links = pd.read_csv("/content/drive/Shareddrives/EDGI - Shared NEW/12_Environmental_Enforcement_Watch/09_Data/Climate_Data_Project/FRS_PROGRAM_LINKS.csv")
coordinates = pd.read_csv("/content/drive/Shareddrives/EDGI - Shared NEW/12_Environmental_Enforcement_Watch/09_Data/Climate_Data_Project/FRS_FACILITIES.csv")

# Join on PGM_SYS_ID
merged = violations.merge(frs_links[["PGM_SYS_ID", "REGISTRY_ID"]],
                          on="PGM_SYS_ID",
                          how="left")
```

```python
# Get list of FRSIDs from ghgrp
ghgrp_ids = ghgrp["FRS Id"].unique()

# Filter merged violations to only those facilities
filtered = merged[merged["REGISTRY_ID"].isin(ghgrp_ids)]

#drop any outside contiguous US (should just be Alaska and Hawaii)
#filtered.drop(filtered[(filtered["STATE_CODE"] == "AK") | (filtered["STATE_CODE"] == "HI")].index, inplace=True)
```

```python
filtered
```

```python
filtered.to_csv("ghgrp_violations.csv", index=False)
```

```python
coordinates_columns = coordinates.columns.tolist()
coordinates_columns
```

```python
located = filtered.merge(coordinates[["FAC_NAME", "FAC_STREET", "FAC_CITY", "FAC_STATE", "FAC_COUNTY", "FAC_EPA_REGION", "REGISTRY_ID", "LATITUDE_MEASURE", "LONGITUDE_MEASURE"]],
                          on="REGISTRY_ID",
                          how="left")
```

```python
located
```

```python
located.to_csv("ghgrp_caa_violations_2023.csv", index=False)
```

```python
ejscreen = gpd.read_file("/content/drive/Shareddrives/EDGI - Shared NEW/12_Environmental_Enforcement_Watch/09_Data/EJScreen_2024_BG_with_AS_CNMI_GU_VI.gdb")
```

```python
ejscreen_columns = ejscreen.columns.tolist()
ejscreen_columns
```

```python
caa_violations = gpd.GeoDataFrame(located, geometry = gpd.points_from_xy(
    x=located['LONGITUDE_MEASURE'], y=located['LATITUDE_MEASURE'], crs=4326)
)
caa_violations.to_crs(crs=4269, inplace=True) # Project to match EJScreen coordinate systems
caa_violations
```

```python
# Spatial join with EJScreen
caa_ejscreen = gpd.sjoin(caa_violations, ejscreen, how="left", predicate='within')
caa_ejscreen
```

```python
air_pollutant_bg = (
    (caa_ejscreen["P_NO2"] > 79) |
    (caa_ejscreen["P_OZONE"] > 79) |
    (caa_ejscreen["P_RSEI_AIR"] > 79) |
    (caa_ejscreen["P_DSLPM"] > 79) |
    (caa_ejscreen["P_PTRAF"] > 79) |
    (caa_ejscreen["P_PM25"] > 79)
)
```
